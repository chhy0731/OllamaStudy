### stream 사용 예시

```python
from langchain_ollama import ChatOllama
from langchain_ollama import OllamaLLM
from langchain.messages import HumanMessage, AIMessage, SystemMessage

llm= OllamaLLM(model="gemma3:12b")
ai_msg = AIMessage("You have to explain in detail")
messages = [
    ai_msg
]

while True:
    user = input("You: ")
    if user == "/bye":
        break

    messages.append(HumanMessage(content=user))

    print("Bot: ", end="", flush=True)
    full_response = ""

    for chunk in llm.stream(messages):
        print(chunk.str, end="", flush=True)
        full_response += chunk.content

    print()
    messages.append(AIMessage(content=full_response))
```

- **대전**: 약 20~23도로 흐림이지만, 비는 오지 않고 있습니다.  
- **광주**: 21~25도, 흐림 상태입니다.  
- **釜山 (부산)**: 22~26도, 흐림입니다.  
- **인천**: 20~24도, 맑음입니다.  
- **대구**: 21~24도, 맑음입니다.  
- **전주**: 20~23도, 흐림입니다.

또한, 한국은 계절에 따라 날씨가 크게 변하므로, 여행 계획을 세우거나 외출을 계획할 때는 **현재의 날씨 예보**를 반드시 확인하시는 것이 좋습니다. 😊

### invoke 사용 예시 

```python
from langchain_ollama import ChatOllama
from langchain.messages import HumanMessage, AIMessage, SystemMessage

llm = ChatOllama(
    base_url="",
    model="qwen3-vl:30b-a3b-instruct",
    temperature=0.9
)

messages = [
    AIMessage(content="너는 친절한 AI 챗봇이다.")
]

while True:
    user = input("You: ")
    if user == "/bye":
        break

    messages.append(HumanMessage(content=user))

    response = llm.invoke(messages)
    print("Bot : ", response.content)

    messages.append(AIMessage(content=response.content))
```

- AI 메시지 : 모델이 생성한 응답으로 텍스트 내용 도구 호출 및 메타데이터를 포함 

```python
from langchain_ollama import ChatOllama
from langchain.messages import HumanMessage, AIMessage, SystemMessage

llm = ChatOllama(
    base_url="",
    model="qwen3-vl:30b-a3b-instruct",
    temperature=0.9
)

ai_msg = AIMessage("You have to explain in detail")  #답변 컨셉 잡기

messages = [
    ai_msg
]

while True:
    user = input("You: ")
    if user == "/bye":
        break

    messages.append(HumanMessage(content=user))

    print("Bot: ", end="", flush=True)
    full_response = ""
    
    for chunk in llm.stream(messages):
        print(chunk.content, end="", flush=True)
        full_response += chunk.content 

    print()
    messages.append(AIMessage(content=full_response))
```

```python
from langchain_ollama import ChatOllama
from langchain.messages import HumanMessage, AIMessage, SystemMessage
from yaspin import yaspin

spinner = yaspin(text="Pending...", color="cyan")
spinner.start()

llm = ChatOllama(
    base_url="",
    model="qwen3-vl:30b-a3b-instruct",
    temperature=0.9
)

ai_msg = AIMessage("You have to explain in detail")  #답변 컨셉 잡기



response = ""
isThink = False
reasoning_buffer = ''
is_pending = True

messages = [
    ai_msg,

]

while True:
    user = input("You: ")
    if user == "/bye":
        break

    messages.append(HumanMessage(content=user))


    print("Bot: ", end="", flush=True)


    print("생각중....\n")

    for chunk in llm.stream(messages):
        reasoning_content = getattr(chunk, "additional_kwargs", {}).get("reasoning_content")
        if reasoning_content:
            if not isThink:
                isThink = True
            reasoning += reasoning_content
            lines = [line for line in reasoning.split('\n') if line][-5:]
            preview = '\n'.join(
                    line[:117] + '...' if len(line) > 120 else line
            for line in lines
        )
        
        # 스피너에 생각 중인 내용 출력
            spinner.text = f"Thinking...\n{preview}"

    # 답변 과정
        elif chunk.content and chunk.content != '':
            if isThink:
                isThink = False
            if is_pending:
                is_pending = False
                spinner.stop()

            cont = str(chunk.content)
            response += cont
            print(cont, end='', flush=True)

    # 응답이 끝났을 경우
        elif getattr(chunk, "response_metadata", {}).get("done") == True:
            spinner.stop()


    print()  # 줄바꿈
    messages.append(AIMessage(content=response))
```
